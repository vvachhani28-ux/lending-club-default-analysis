## Data Analysis

Let's now move to data analysis. To start with, let's understand the objective of the analysis clearly and identify the variables that we want to consider for analysis. 

The objective is to identify predictors of default so that at the time of loan application, we can use those variables for approval/rejection of the loan. Now, there are broadly three types of variables - 1. those which are related to the applicant (demographic variables such as age, occupation, employment details etc.), 2. loan characteristics (amount of loan, interest rate, purpose of loan etc.) and 3. Customer behaviour variables (those which are generated after the loan is approved such as delinquent 2 years, revolving balance, next payment date etc.).

Now, the customer behaviour variables are not available at the time of loan application, and thus they cannot be used as predictors for credit approval. 

Thus, going forward, we will use only the other two types of variables.



```python
behaviour_var =  [
  "delinq_2yrs",
  "earliest_cr_line",
  "inq_last_6mths",
  "open_acc",
  "pub_rec",
  "revol_bal",
  "revol_util",
  "total_acc",
  "out_prncp",
  "out_prncp_inv",
  "total_pymnt",
  "total_pymnt_inv",
  "total_rec_prncp",
  "total_rec_int",
  "total_rec_late_fee",
  "recoveries",
  "collection_recovery_fee",
  "last_pymnt_d",
  "last_pymnt_amnt",
  "last_credit_pull_d",
  "application_type"]
behaviour_var
```

A good way to quantify th effect of a categorical variable on default rate is to see 'how much does the default rate vary across the categories'. 

Let's see an example using annual_inc as the categorical variable.

```python
# variation of default rate across annual_inc
df.groupby('annual_inc').loan_status.mean().sort_values(ascending=False)
```

```python
# one can write a function which takes in a categorical variable and computed the average 
# default rate across the categories
# It can also compute the 'difference between the highest and the lowest default rate' across the 
# categories, which is a decent metric indicating the effect of the varaible on default rate

def diff_rate(cat_var):
    default_rates = df.groupby(cat_var).loan_status.mean().sort_values(ascending=False)
    return (round(default_rates, 2), round(default_rates[0] - default_rates[-1], 2))

default_rates, diff = diff_rate('annual_inc')
print(default_rates) 
print(diff)
```

Thus, there is a 6% increase in default rate as you go from high to low annual income. We can compute this difference for all the variables and roughly identify the ones that affect default rate the most.

```python
# filtering all the object type variables
df_categorical = df.loc[:, df.dtypes == object]
df_categorical['loan_status'] = df['loan_status']

# Now, for each variable, we can compute the incremental diff in default rates
print([i for i in df.columns])
```

```python
# storing the diff of default rates for each column in a dict
d = {key: diff_rate(key)[1]*100 for key in df_categorical.columns if key != 'loan_status'}
print(d)
```

## Conclusions

Based on the univariate, segmented, and difference-in-default-rate analysis above, the variables most strongly associated with default are, in order of impact:

1. **Loan grade / interest rate** — default rate rises from ~5.8% (Grade A) to ~33.6% (Grade G), the largest effect of any variable studied.
2. **Loan term** — 60-month loans default at more than double the rate of 36-month loans (25.1% vs 10.9%).
3. **Loan purpose** — `small_business` loans default at 27.0%, nearly double the overall average of 14.4%, and this elevated risk persists across grades (see the purpose × grade cross-tab).
4. **Annual income and DTI** — both move default rate in the expected direction (higher income → lower default, higher DTI → higher default), but the effect size is smaller (roughly 5-6 percentage points across bands) than grade, rate, or term.

These findings are consistent across the univariate breakdowns, the segmented (purpose × grade) analysis, and the diff-rate summary computed above.

## Recommendations

1. **Tighten underwriting or re-price Grade E/F/G loans.** These grades default at 4-6x the rate of Grade A. If continuing to originate them, ensure pricing (interest rate) adequately compensates for the higher expected loss.
2. **Apply additional scrutiny to `small_business` and `renewable_energy` purpose loans.** Their elevated default rate holds across grades, suggesting purpose carries risk signal independent of the grade already assigned — consider extra documentation requirements or lower exposure limits for these categories.
3. **Reconsider offering 60-month terms for lower grades.** The combination of a weak grade and a long repayment term compounds risk substantially, and is one of the highest-risk combinations in the dataset.
4. **Use income and DTI to fine-tune pricing within a grade, not as standalone approval filters.** Their effect on default rate, while real, is modest compared to grade, interest rate, and term.
5. **Interpret year-over-year default rate trends with caution.** Loans issued more recently (e.g. 2011) have had less time to default than older loans (e.g. 2007), so raw year-over-year comparisons can understate the eventual default rate of newer vintages.

### Limitations
- This is a descriptive/exploratory analysis identifying association, not a causal or predictive model.
- Grade itself is partially derived from some of the same underlying risk factors examined here, so grade and other variables are not fully independent.
- No loan-age/vintage adjustment was applied to the time-trend analysis; this would be a natural next step before drawing firm conclusions about risk trending up or down over time.
